```python
#Slicing lists
seq = [7, 2, 3, 7, 5, 6, 0, 1]
b=seq[1:5]
b
```

```python
#Slicing can be used to substitute values
print(seq)
seq[3:5] = [6, 3]
seq
```

```python
print(seq[:5])
print(seq[3:])
```

```python
#Slicing can use negative indices
print(seq)
print(seq[-4:])
print(seq[-6:-2])
```

```python
#The Dictionary or dict data structure
#A dictionary stores a collection of key-value pairs, where key and value are Python objects. 
#Each key is associated with a value so that a value can be conveniently retrieved, inserted, 
#modified, or deleted given a particular key.
d1 = {"a": "some value", "b": [1, 2, 3, 4]}
d1
```

```python
print(d1["a"])
print(d1["b"])
```

```python
d1[7] = "new text"
d1
```

```python
print(d1)
print(list(d1.keys()))
print(list(d1.values()))
```

```python
#Updating a dictionary
d1.update({"b": "foo", "c": 12})
d1
```

```python
#Expunging a dictionary
d1.pop("a")
d1
```

```python
#Functions
#Python allows you to build functions as you need them
#Here is a simple example
def newabsolute(x,y):
    if x+y>0:
        return x+y
    else:
        return -(x+y)
```
