```python
#An if statement can be optionally followed by one or more elif blocks and a catchall
#else block if all of the conditions are False.
#If any of the conditions are True, no further elif or else blocks will be reached.

x=3.7
if x < 0:
    print("It's negative")
elif x == 0:
    print("Equal to zero")
elif 0 < x < 5:
    print("Positive but smaller than 5")
else:
    print("Positive and larger than or equal to 5")
```

```python
#for loops
#Check for indentation and :
a = [1, 4, 3]
total = 0
for value in a:
    total += value
total
```

```python
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
total
```

```python
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
total_until_5 = 0
for value in sequence:
    if value == 0:
        break
    total_until_5 += value
total_until_5
```
