```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
np.set_printoptions(legacy='1.21')
```

```python
#Uploading data with pandas
nas = pd.read_csv('nasdaq 2026.csv')
```

```python
nasvalue=nas['Close/Last']
```

```python
nasdaqreturn = []
for i in range(0, len(nasvalue)-1):
    current_price = nasvalue[i]
    previous_price = nasvalue[i+1]
    percentage_return = ((current_price - previous_price) / previous_price) * 100
    nasdaqreturn.append(percentage_return)
```

```python
len(nasdaqreturn)
```

```python
#We now do the same for Apple
apple = pd.read_csv('Apple 2026.csv')
apvalue=apple['Close/Last']
apreturn = []
for i in range(0, len(apvalue)-1):
    current_price = apvalue[i]
    previous_price = apvalue[i+1]
    percentage_return = ((current_price - previous_price) / previous_price) * 100
    apreturn.append(percentage_return)
len(apreturn)
```

```python
#Computing correlation coefficient
data = {'apple': apreturn,
        'nadsdaq': nasdaqreturn}
applenasdaqindices = pd.DataFrame(data)
```

```python
covariance_matrix = applenasdaqindices.cov()
covariance_matrix
```

```python
# Calculate the correlation matrix for the DataFrame
correlation_matrix = applenasdaqindices.corr()
print(correlation_matrix)
```

```python
a=np.random.randn(1000)
b=np.random.randn(1000)
c=np.random.randn(1000)
```

```python
fig, ax = plt.subplots()
ax.plot(a.cumsum(), color="black", label="one")
ax.plot(b.cumsum(), color="red", linestyle="dashed", label="two")
ax.plot(c.cumsum(), color="green", linestyle="dotted", label="three")
ax.legend()
ax.set(title="Three random walks plot")
ticks = ax.set_xticks([0, 250, 500, 750, 1000])
labels = ax.set_xticklabels(["one", "two", "three", "four", "five"],rotation=30, fontsize=9)
```
