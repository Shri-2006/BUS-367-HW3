```python
#An Example. Simulating a random walk.
import random
#This imports Python's built-in random module. The module generating random numbers. 
#We will use it below to decide whether the walk moves +1 or −1 at each step.
```

```python
position = 0
#This creates a variable (called 'position', but any name will do).
#It is assigned the value 0, the initial location of the random walker.

walk = [position]
#We now create a list (notice the [] format). For the moment the list just contains the initial 
#position, i.e. 0.

walk
```

```python
random.randint(0, 1)
#A random number generator. It generates 0 or 1,
#each with probability 0.5
```

```python
random.randint(2, 3)
#It can be used for any pair of integers.
```

```python
random.randint?
```

```python
random.randint(0,2)
```

```python
nsteps = 1000
#This define the variable that we will be used 
#to establish the length of the walk.

for _ in range(nsteps):
#The underscore _ is used because we don't actually need the 
#loop counter. We only care about repeating the operation 1,000 times.
    step = 1 if random.randint(0, 1) else -1
    #This defines 'step' as equal to 1 if random.randint(0, 1)
    #is 1, and equal to -1 otherwise.
    position += step
    #This redefines the variable 'position' adding 'step'.
    #So, at each round 'position' is either increased by 1 or decreased by 1 
    walk.append(position)
    #Redifenes the list walk adding the new position at each round
```

```python
walk[0:20]
```

```python
import matplotlib.pyplot as plt
#matplotlib is the main module to generate pictures. 
#More on this later

plt.plot(walk)
#Plots the random walk and generates picture.
```

```python
plt.plot(walk)
```

```python
#We can plot any portion of the random walk
plt.plot(walk[:30])
```

```python
plt.plot(walk[30:70])
```

```python
#Producing a better graph
plt.plot(walk)
plt.xlabel("Step")
plt.ylabel("Position")
plt.title("One-Dimensional Random Walk")
plt.show()
```

```python
#Evolution of a stock price
price = 100
#Initial price

prices = [price]
#Creates list with all prices

nsteps = 50
#Number of periods 
```

```python
#Creating the price path.
for _ in range(nsteps):
    if random.randint(0, 1):
        price *= 1.002
    else:
        price *= 0.999    
    prices.append(price)
```

```python
prices[10:20]
```

```python
plt.plot(prices)
plt.xlabel("Time")
plt.ylabel("Price")
plt.title("Evolution of stock price, +0.2% or -0.1% equal probability")
plt.show()
```
