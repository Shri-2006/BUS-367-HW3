```python
#Reindexing
#You can create a new object with the values rearranged to align with the new index
obj = pd.Series([4.5, 7.2, -5.3, 3.6], index=["d", "b", "a", "c"])
obj
```

```python
#If we put an index name not present in the original formulation,
#we get NaN
obj2 = obj.reindex(["a", "b", "c", "d", "e"])
obj2
```

```python
#Reindexing can also be used for dataframes
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),index=["a", "c", "d"],columns=["Ohio", "Texas", "California"])
frame
```

```python
frame2 = frame.reindex(index=["a", "b", "c", "d"])
frame2
```

```python
frame2["Utah"]=[0,0,0,0]
frame2
```

```python
#When you add objects, if any index pairs are not the same, the respective
#index in the result will be the union of the index pairs
s1 = pd.Series([7.3, -2.5, 3.4, 1.5], index=["a", "c", "d", "e"])
s2 = pd.Series([-2.1, 3.6, -1.5, 4, 3.1], index=["a", "c", "e", "f", "g"])
print(s1)
print(s2)
```

```python
#Adding the two series will give NaN when there is no shared index.
sumseries=s1+s2
sumseries
```

```python
type(sumseries)
```

```python
#Similar for Dataframes
df1 = pd.DataFrame(np.arange(9.).reshape((3, 3)), columns=list("bcd"), index=["Ohio", "Texas", "Colorado"])
df2 = pd.DataFrame(np.arange(12.).reshape((4, 3)), columns=list("bde"), index=["Utah", "Ohio", "Texas", "Oregon"])
print(df1)
print(df2)
```

```python
#Only Ohio and Texas belong to both DataFrames. Also, only columns "b" and "d"
#belong to both. In all other cases we get NaN
df1+df2
```

```python
#This is another method to sum dataframes. It automatically fills with an assigned value
#when a pair only appears in one dataframe. When the pair is not present
#in both, then NaN is assigned.

df1.add(df2, fill_value=0)
```

```python
#Operations between Series and Dataframes
arr = np.arange(12.).reshape((3, 4))
arr
```

```python
#Recover the value in row 0, column 2
arr[0,2]
```

```python
#Recover column 0 
arr[:,0]
```

```python
#Recover row 1
arr[1,:]
```

```python
#Simpler way to recover a row
arr[1]
```

```python
#Subtracts the same row (in this case subtract row 0
#from itself and all other rows
arr-arr[0]
```

```python
#Operation between DataFrames and Series are similar
frame = pd.DataFrame(np.arange(12.).reshape((4, 3)), columns=list("bde"), 
                     index=["Utah", "Ohio", "Texas", "Oregon"])
frame
```

```python
#The command .iloc[] can be used to select rows and 
#columns from a DataFrame by their integer position.

utah = frame.iloc[0]
#selects row 0 from the dataset frame
utah
```

```python
type(utah)
```

```python
#Selects column 1 (the one labeled ''d'')
dcolumn=frame.iloc[:,1]
dcolumn
```

```python
#By default, arithmetic between DataFrame and Series matches the index of the Series
#on the columns of the DataFrame.
frame-utah
```
